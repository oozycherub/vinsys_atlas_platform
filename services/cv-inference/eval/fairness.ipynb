# Face-verification fairness (PRD §4.1, docs/prompts/10 §8)

Error-rate parity across Fitzpatrick skin-type cohorts at the production threshold (cosine ≥ 0.45). Release gate: the worst cohort's FNMR and FMR are at most 1.5× the best cohort's. The pairs CSV comes from the consented, fairness-balanced evaluation set scored by the production ArcFace model; protected attributes never leave this offline set.


```python
import csv
from pathlib import Path

import fairness

PAIRS = Path('pairs.csv')  # cohort,same_person,cosine
rows = list(csv.DictReader(PAIRS.open(newline='')))
report = fairness.evaluate(rows)
report['passed'], report['fnmr_ratio'], report['fmr_ratio']
```

```python
# Threshold sensitivity: parity across thresholds around the production value.
for t in (0.35, 0.40, 0.45, 0.50, 0.55):
    r = fairness.evaluate(rows, threshold=t)
    print(f"{t:.2f}  FNMR ratio {r['fnmr_ratio']:.2f}  FMR ratio {r['fmr_ratio']:.2f}  passed {r['passed']}")
```

```python
for c in report['cohorts']:
    print(f"{c['cohort']:>8}  genuine {c['genuine']:>5}  impostor {c['impostor']:>5}  FNMR {c['fnmr']:.3%}  FMR {c['fmr']:.3%}")
```
